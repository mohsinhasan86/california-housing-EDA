```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/california_housing_cleaned.csv')

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
numeric_cols = [
    'housing_median_age',
    'total_rooms',
    'population',
    'households',
    'median_income',
    'median_house_value',
    'rooms_per_household',
    'bedrooms_per_room',
    'population_per_household'
]

df[numeric_cols].hist(figsize=(15, 12), bins=30)

plt.suptitle("Distribution of Key Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
for col in numeric_cols:
    print("\n", "="*50)
    print(col)
    print("Mean  :", round(df[col].mean(), 2))
    print("Median:", round(df[col].median(), 2))
    print("Min   :", round(df[col].min(), 2))
    print("Max   :", round(df[col].max(), 2))
```

```python
display(df[numeric_cols].quantile([0.01, 0.25, 0.50, 0.75, 0.99]).T)
```

```python
display(df[numeric_cols].quantile([0.01, 0.25, 0.50, 0.75, 0.99]).T)
```

```python
print("Ocean Proximity Categories:")
print(df['ocean_proximity'].value_counts())
```

```python
category_summary = df.groupby('ocean_proximity').agg(
    records=('respondent_id', 'count'),
    avg_income=('median_income', 'mean'),
    median_income=('median_income', 'median'),
    avg_house_value=('median_house_value', 'mean'),
    median_house_value=('median_house_value', 'median')
).round(2)

display(category_summary)
```

```python
category_summary['avg_house_value'].plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title("Average House Value by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Average House Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Observation: Average house value varies across ocean-proximity categories. Inland records have a lower average house value than the coastal/bay categories in this sample. The Island category should be interpreted cautiously because it contains only 19 records.

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['median_income'],
    df['median_house_value'],
    alpha=0.5
)

plt.title("Median Income vs Median House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")

plt.tight_layout()
plt.show()
```

The scatter plot suggests a positive association between median income and median house value in this dataset."

```python
correlation = df['median_income'].corr(df['median_house_value'])

print("Correlation between Median Income and Median House Value:",
      round(correlation, 3))
```

The correlation coefficient between median income and median house value is 0.942, indicating a very strong positive linear association between the two variables. However, correlation does not imply causation.

```python
top_house_values = df.sort_values(
    by='median_house_value',
    ascending=False
)[[
    'median_income',
    'median_house_value',
    'ocean_proximity',
    'housing_median_age'
]].head(10)

display(top_house_values)
```

The house-value variable appears to have an upper cap at 500001, as multiple observations share the exact maximum value. This may limit the ability to distinguish properties above this threshold.”

```python
capped_values = (df['median_house_value'] == 500001).sum()

print("Number of records at 500001:", capped_values)

print("Percentage of records:",
      round((capped_values / len(df)) * 100, 2), "%")
```

“1.6% of observations have a median house value of exactly 500001, suggesting a possible upper-value cap. These observations may limit our ability to distinguish properties whose actual values could be above this threshold.”

```python
top_bedroom_ratio = df.sort_values(
    by='bedrooms_per_room',
    ascending=False
)[[
    'bedrooms_per_room',
    'total_rooms',
    'total_bedrooms',
    'population',
    'median_income',
    'median_house_value',
    'ocean_proximity'
]].head(10)

display(top_bedroom_ratio)
```

Several observations contain unusually high bedrooms_per_room ratios, driven by the same imputed-looking total_bedrooms value of 1015.54 across different records. These observations should be investigated before using the ratio for downstream analysis.

```python
count_1015 = (df['total_bedrooms'] == 1015.541856).sum()

print("Records with total_bedrooms = 1015.541856:", count_1015)
```

```python
display(
    df.loc[df['bedrooms_per_room'].idxmax(), [
        'total_rooms',
        'total_bedrooms',
        'bedrooms_per_room',
        'population',
        'households',
        'median_income',
        'median_house_value',
        'ocean_proximity'
    ]]
)
```

An extreme value was identified in bedrooms_per_room: 15.39, where total bedrooms (1,015.54) greatly exceed total rooms (66). This observation should be investigated as a potential data-quality anomaly before using this derived variable for further analysis.”

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['median_income'],
    df['median_house_value'],
    alpha=0.5
)

plt.title("Median Income vs Median House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")

plt.annotate(
    "Correlation = 0.942",
    xy=(8, 400000),
    fontsize=12
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

df.groupby('ocean_proximity')['median_house_value'].mean().sort_values().plot(
    kind='bar'
)

plt.title("Average House Value by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Average House Value")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['median_house_value'],
    bins=30
)

plt.title("Distribution of Median House Value")
plt.xlabel("Median House Value")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['housing_median_age'],
    df['median_house_value'],
    alpha=0.5
)

plt.title("Housing Median Age vs Median House Value")
plt.xlabel("Housing Median Age")
plt.ylabel("Median House Value")

plt.tight_layout()
plt.show()
```

# EDA Findings & Conclusion

## Key Findings

### 1. Strong relationship between income and house value
The correlation between median income and median house value is 0.942, indicating a very strong positive linear association. In general, areas with higher median income tend to have higher median house values in this dataset.

However, correlation does not imply causation.

### 2. Possible upper cap in house values
32 out of 2,000 observations (1.6%) have a median house value of exactly 500001.

The repeated maximum value suggests that the house-value variable may have an upper cap. This should be considered when interpreting high-value properties and performing further analysis.

### 3. Differences across ocean-proximity categories
Average house values vary across ocean-proximity categories.

INLAND areas have a lower average house value than the major coastal-related categories in this sample. The ISLAND category has the highest average value, but it contains only 19 observations, so this result should be interpreted cautiously.

These differences represent associations in the dataset and do not establish causation.

### 4. Potential data-quality anomaly
An unusually high bedrooms_per_room value of 15.39 was identified.

The corresponding observation contains 66 total rooms and 1,015.54 total bedrooms. This is an unusual combination and should be investigated before using bedrooms_per_room for downstream analysis.

## Final Conclusion

The exploratory data analysis revealed several important patterns in the California housing dataset.

The strongest observed relationship is between median income and median house value, with a correlation of 0.942. House values also vary across ocean-proximity categories. In addition, the analysis identified a possible upper cap in median house value and a potential data-quality anomaly in bedrooms_per_room.

These findings provide useful hypotheses for further analysis, but they should not be interpreted as causal relationships without additional investigation.

Overall, the EDA helped identify important distributions, relationships, segmentation patterns, and data-quality issues that can guide the next stage of analysis.
